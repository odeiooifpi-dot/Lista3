Questão 1

```python

notas = []
for i in range(5) :
    nota = float(input("Digite a nota: "))
    notas.append(nota)
soma = 0
for nota in notas :
    soma += nota
media = soma / len(notas)
acima_ou_igual = 0
for nota in notas :
    if nota >= media :
        acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

```python
numeros = []
for i in range(8):
    numero = int(input("Digite um número: "))
    numeros.append(numero)
maior = numeros[0]
menor = numeros[0]
for i in range(1, 8):
    if numeros[i] > maior:
        maior = numeros[i]
    if numeros[i] < menor:
        menor = numeros[i]
pos_maior = numeros.index(maior)
pos_menor = numeros.index(menor)
print("Lista:", numeros)
print("Maior:", maior)
print("Posição do maior:", pos_maior)
print("Menor:", menor)
print("Posição do menor:", pos_menor)
```

Questão 12

```python
estoque = {
    "Arroz": 20,
    "Feijão": 15,
    "Leite": 10,
    "Café": 8
}
produto = input("Digite o nome do produto: ")
quantidade = int(input("Digite a quantidade vendida: "))
if produto in estoque:
    if quantidade <= estoque[produto]:
        estoque[produto] = estoque[produto] - quantidade
        print("Venda realizada com sucesso!!")
    else:
        print("Estoque insuficiente")
else:
    print("Produto não encontrado")
print("Estoque final:")
print(estoque)
```

Questão 15

```python
Lista: Foi usada para armazenar os 5 itens do estoque
Dicionário: Foi usado para guardar as informações de cada item, como os códigos, nomes, quantidades e preços
Tupla: Foi usada para guardar o resumo final com a quantidade de itens e o seu valor total

estoque = []
for i in range(5):
    item = {}
    item["codigo"] = int(input("Codigo: "))
    item["nome"] = input("Nome: ")
    item["quantidade"] = int(input("Quantidade: "))
    item["preco"] = float(input("Preco: "))
    estoque.append(item)
print("\nItens:")
for item in estoque:
    print(item)
total = 0
maior = 0
item_maior = ""
for item in estoque:
    valor = item["quantidade"] * item["preco"]
    total = total + valor
    if valor > maior:
        maior = valor
        item_maior = item["nome"]
resumo = (len(estoque), total)
print("\nValor total:", total)
print("Maior valor:", maior)
print("Item de maior valor:", item_maior)
print("Resumo:", resumo)
```
